```python
import numpy as np

# Example array
data = np.array([10, 2, 38, 27, 38, 23, 21])
data2 = np.array([10, 2, 38, 27, 38, 23, 21, 22])

# Calculate the median
median_value = np.median(data)

print("Median:", median_value)
print("Median1", np.median(data2))
```

```python
import numpy as np

# Example 2D array
data = np.array([[10, 7, 4], [3, 2, 1], [6, 8, 5]])

# Calculate the median along the first axis (columns)
median_value_axis0 = np.median(data, axis=0)

# Calculate the median along the second axis (rows)
median_value_axis1 = np.median(data, axis=1)

print("Median along axis 0:", median_value_axis0)
print("Median along axis 1:", median_value_axis1)
```
